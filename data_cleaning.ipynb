A variable is treated as 'mostly present' when under 50% of values are missing. The distribution of missing shares is concentrated near 0 and 1, but 488 of 3815 rows fall between 0.05 and 0.95, so still to check: how much the classification changes at thresholds 0.2 / 0.5 / 0.8.

```python
# Clean the marker to True/False and classify
both['collected_clean'] = both['collected'].astype(str).str.strip().str.lower() == 'x'

# Burundi 2020 B is the stray newline in the source sheet for site_longitude. The data is 69% filled, so treated as collected
mask_sitelong = (both['country'] == 'Burundi') & (both['year'] == 2020) & (both['season'] == 'B') & (both['variable'] == 'site_longitude')
both.loc[mask_sitelong, 'collected_clean'] = True
```

```python
# Classify each row by comparing the marker with the missing share.
THRESHOLD = 0.5

import numpy as np

conditions = [
    both['collected_clean'] & (both['missing_share'] < THRESHOLD),
    both['collected_clean'] & (both['missing_share'] < 1),
    both['collected_clean'] & (both['missing_share'] == 1),
    ~both['collected_clean'] & (both['missing_share'] == 1),
    ~both['collected_clean'] & (both['missing_share'] < 1),
]
labels = ['consistent', 'missingness', 'possible documentation error',
          'consistent', 'contradiction']

both['status'] = np.select(conditions, labels, default='unclassified')
```

```python

QUALS = ["north", "south", "east", "west", "upper", "lower", "low", "central", "centre"]
ALIAS = {"low": "lower", "centre": "central"}

pat_start = re.compile(rf"^({'|'.join(QUALS)})")
pat_end = re.compile(rf"({'|'.join(QUALS)})$")


def clean(s):
    return re.sub(r"[^a-z0-9]", "", str(s).lower())


def split_quals(key):
    quals = set()
    changed = True

    while changed:
        changed = False

        for p in (pat_start, pat_end):
            m = p.search(key)

            if m and len(key) - len(m.group(1)) >= 3:
                quals.add(ALIAS.get(m.group(1), m.group(1)))
                key = p.sub("", key, count=1)
                changed = True

    return key, frozenset(quals)


# Work only with non-missing site_clean values
work = kenya.loc[
    kenya["site_clean"].notna(),
    ["site_clean"]
].copy()

work["site_clean"] = work["site_clean"].astype(str)

work = work[
    work["site_clean"].str.strip() != ""
]


# Create normalized matching key
work["key1"] = work["site_clean"].map(clean)

# Remove values that contain only punctuation
work = work[work["key1"] != ""]


# Separate base name and geographic qualifier
work["base"], work["quals"] = zip(
    *work["key1"].map(split_quals)
)


# Create final grouping key
work["key2"] = (
    work["base"]
    + "|"
    + work["quals"].map(lambda q: ",".join(sorted(q)))
)


# Choose the most frequent spelling as the canonical name
canon = (
    work.groupby("key2")["site_clean"]
    .agg(lambda s: s.value_counts().index[0])
)


# Assign canonical name
work["site_std"] = work["key2"].map(canon)


# Put standardized values back into Kenya dataframe
kenya["site_std"] = work["site_std"]
kenya["site_key"] = work["key2"]


# Review groups where multiple spellings were combined
dups = (
    work.groupby("key2")["site_clean"]
    .agg(lambda s: sorted(set(s)))
)

dups = dups[dups.map(len) > 1]

print(len(dups), "groups to review")
print(dups.to_string())
```

```python
kenya["growing_days"] = (kenya["harvest_date"] - kenya["plant_date"]).dt.days

# statistical outliers relative to the Kenya growing-period distribution,
q1 = kenya["growing_days"].quantile(0.25)
q3 = kenya["growing_days"].quantile(0.75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower:", lower)
print("Upper:", upper)


```

```python
outliers = kenya[
    (kenya["growing_days"] < lower) |
    (kenya["growing_days"] > upper)
]

outliers[
    ["unique_id", "year", "plant_date", "harvest_date",
     "growing_days", "crop"]
].sort_values("growing_days")
```

```python
MIN_D, MAX_D = lower, upper

# Backups (only once, before any change)
if "plant_date_orig" not in kenya:
    kenya["plant_date_orig"] = kenya["plant_date"]
    kenya["harvest_date_orig"] = kenya["harvest_date"]

# no harvest date if yield is null or 0 in BOTH yield columns
no_yield = (
    (kenya["yield_kg_pa"].isna() | (kenya["yield_kg_pa"] == 0)) &
    (kenya["yield_kg_ph"].isna() | (kenya["yield_kg_ph"] == 0))
)
kenya.loc[no_yield, "harvest_date"] = pd.NaT

kenya["growing_days"] = (kenya["harvest_date"] - kenya["plant_date"]).dt.days
kenya["date_fix"] = pd.NA
kenya.loc[no_yield & kenya["harvest_date_orig"].notna(), "date_fix"] = "harvest_removed_no_yield"


def shift(dt, target_year):
    return dt + pd.DateOffset(years=int(target_year - dt.year))


def fix_row(r):
    p, h, y = r["plant_date"], r["harvest_date"], r["year"]

    if pd.isna(p) or pd.isna(h) or pd.isna(y):
        return p, h, pd.NA

    # Planting year is after survey year, harvest is in survey year
    if p.year > y and h.year == y and 2 <= p.month <= 5:
        return shift(p, y), h, "plant_year->year"

    # Harvest year is before survey year, planting is in survey year
    if h.year < y and p.year == y:
        return p, shift(h, y), "harvest_year->year"

    # December planting with harvest between May and September
    # in the same calendar year
    if p.month == 12 and h.year == p.year and 5 <= h.month <= 9:
        return shift(p, y - 1), h, "plant_year->year-1"

    return p, h, "unresolved"


# Only rows that still have a harvest date and an out-of-range duration
bad_idx = kenya.index[
    ~no_yield &
    ((kenya["growing_days"] < MIN_D) | (kenya["growing_days"] > MAX_D))
]

for i in bad_idx:
    p, h, cause = fix_row(kenya.loc[i])

    if pd.isna(cause):
        continue

    new_days = (h - p).days

    if cause != "unresolved" and MIN_D <= new_days <= MAX_D:
        kenya.loc[i, ["plant_date", "harvest_date"]] = [p, h]
        kenya.loc[i, "date_fix"] = cause
    else:
        kenya.loc[i, ["plant_date", "harvest_date"]] = pd.NaT
        kenya.loc[i, "date_fix"] = "unresolved->NaT"


# Recalculate after cleaning
kenya["growing_days"] = (kenya["harvest_date"] - kenya["plant_date"]).dt.days
kenya["plant_year"] = kenya["plant_date"].dt.year
kenya["harvest_year"] = kenya["harvest_date"].dt.year

print("Bounds:", MIN_D, "to", MAX_D)
print("\nHarvest dates removed for no yield:", (kenya["date_fix"] == "harvest_removed_no_yield").sum())
print("\nDate fixes:")
print(kenya["date_fix"].value_counts())
print("\nGrowing days:")
print(kenya["growing_days"].describe())
```
